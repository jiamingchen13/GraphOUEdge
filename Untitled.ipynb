```python
import os
import glob
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cvxpy as cp
from itertools import combinations
from datetime import datetime
from data_process import *
from mcar.estimate import *
from mcar.simulate import *
from graphOUedge import GraphOUEdge
from numpy.lib.stride_tricks import sliding_window_view
from statsmodels.tsa.api import VAR
from statsmodels.tsa.ar_model import AutoReg

# Initialization

# Tickers
TICKERS = ['SPY', 'JPM', 'KO', 'MSFT']

# Maximum neighbor for each lag
R_L = [1]

# Number of covariate
C = 1

# Aggregation interval for realized covariance
rule = '5min'

# Set jumps to be a compounded Poisson process
jump_activity = 'finite'

# Forecast steps
forecast_step = 5
```

```python
# Training set
for TICKER in TICKERS:
    input_path = './data/LOBSTERdata/_data_dwn_16_88__' + TICKER + '_2023-01-01_2025-07-31_1'
    log_path = './data/LOBSTERdata/log/train/' + TICKER + '_processing_logs'
    os.makedirs(log_path, exist_ok=True)
    output_path = './data/LOBSTERdata/output/train/' + TICKER
    os.makedirs(output_path, exist_ok=True)
    multiprocess_orderbooks(TICKER=TICKER, input_path=input_path, output_path=output_path, log_path=log_path, check_for_processed_data=True)
    
aggregate_assets(TICKERS, './data/LOBSTERdata/output/train/', './data/LOBSTERdata/output/train/aggregate/')
```

```python
# Testing set
for TICKER in TICKERS:
    input_path = './data/LOBSTERdata/_data_dwn_16_88__' + TICKER + '_2025-08-01_2025-08-31_1'
    log_path = './data/LOBSTERdata/log/test/' + TICKER + '_processing_logs'
    os.makedirs(log_path, exist_ok=True)
    output_path = './data/LOBSTERdata/output/test/' + TICKER
    os.makedirs(output_path, exist_ok=True)
    multiprocess_orderbooks(TICKER=TICKER, input_path=input_path, output_path=output_path, log_path=log_path, check_for_processed_data=True)
    
aggregate_assets(TICKERS, './data/LOBSTERdata/output/test/', './data/LOBSTERdata/output/test/aggregate/')
```

```python
# Load empirical data
data = np.load('./data/LOBSTERdata/output/train/aggregate/cov.npz')
index = pd.to_datetime(data["timestamp"])
columns = [key for key in data.files if key != "timestamp"]
cov_data = pd.DataFrame({col: data[col] for col in columns}, index=index)
data.close()

# Derive network structure
Adj = line_graph(cov_data, TICKERS, rule)

# Number of edges
K = Adj[0].shape[0]

# Derive the adjacency matrix for edge graph
Adj_L = adjacency_matrices(Adj[0], max(R_L))

# Compute the edge weight
W_C = edge_weight(Adj_L)

# Training data
df_cov = cov_data.resample(rule).sum()
df_cov = df_cov.loc[df_cov.index.intersection(cov_data.index)]
Y = df_cov[Adj[1]].T

# Set time grids
n = int(np.ceil((Y.shape[1] - 1) ** (1 / 7)))
P_ = np.linspace(0, (Y.shape[1] - 1), Y.shape[1])
Q_ = np.append(P_[::n ** 4], P_[-1]) if (Y.shape[1] - 1) % n ** 4 != 0 else P_[::n ** 4] 

# Testing data
test_data = np.load('./data/LOBSTERdata/output/test/aggregate/cov.npz')
test_index = pd.to_datetime(test_data["timestamp"])
test_columns = [key for key in test_data.files if key != "timestamp"]
test_cov_data = pd.DataFrame({col: test_data[col] for col in test_columns}, index=test_index)
test_data.close()

df_test_cov = test_cov_data.resample(rule).sum()
df_test_cov = df_test_cov.loc[df_test_cov.index.intersection(test_cov_data.index)]
test_Y = df_test_cov[Adj[1]].T
test_set = sliding_window_view(test_Y, (K, forecast_step))[0]
test_size = test_set.shape[0]
```

```python
# Graph OU Edge Model

X = GraphOUEdge(R_L, K, C, Adj_L, W_C, 'data\\LOBSTERdata\\output\\train\\aggregate', 'plots')
aic = X.estimate_unknown_Levy(Y, f'{jump_activity}') * -2 + 2 * (len(R_L) + sum(R_L) * C)
b = np.load(f'./data/LOBSTERdata/output/train/aggregate/MCAR_estimates_b_{jump_activity}.npy')
Sigma = np.load(f'./data/LOBSTERdata/output/train/aggregate/MCAR_estimates_Sigma_{jump_activity}.npy')
F_mean = np.load(f'./data/LOBSTERdata/output/train/aggregate/MCAR_estimates_F_mean_{jump_activity}.npy')
F_squared = np.load(f'./data/LOBSTERdata/output/train/aggregate/MCAR_estimates_F_squared_{jump_activity}.npy')

estimates = np.zeros((test_size, len(R_L) + sum(R_L) * C))
forecast_grOU = np.zeros((test_size, K, forecast_step))
df_grOU = np.array(Y.copy())
jump_F = scipy.stats.multivariate_normal(mean=F_mean, cov=(F_squared - np.outer(F_mean, F_mean)), allow_singular=True)
alphas = []
betas = []

for i in range(test_size):
    window_data = df_grOU[:, i:]
    theta_hat, HH, H = estimate_grCAR_drift(window_data, Adj_L, W_C, len(R_L), R_L, C, P_, Q_, b, None, Sigma, with_cov=True)

    # Fetch parameters
    idx = 0
    alpha_pred = [] 
    beta_pred = []
    for r in R_L:
        alpha_pred.append(theta_hat[idx])
        beta_pred.append(theta_hat[(idx + 1):(idx + 1 + r * C)])
        idx += 1 + r * C
    X.alpha_set(alpha_pred)
    X.beta_set(beta_pred)

    try:
        Q_Q = X.Q_matrix()
    except:
        x = cp.Variable(len(R_L) + sum(R_L))

        # Objective function
        objective = cp.Minimize(-H.T @ x + 0.5 * cp.quad_form(x, HH))
    
        # Constraint: the symmetric part of Q_Q is negative definite
        epsilon = 0
        constraints = [estimator_to_QQ_cp(x, R_L, C, W_C, Adj_L, K) << -epsilon * np.eye(len(R_L) * K)]
    
        # Problem setup
        prob = cp.Problem(objective, constraints)
        prob.solve(solver=cp.CVXOPT, abstol=1e-6, reltol=1e-6, feastol=1e-6)
        theta_hat = x.value
        idx = 0
        alpha_pred = [] 
        beta_pred = []
        for r in R_L:
            alpha_pred.append(theta_hat[idx])
            beta_pred.append(theta_hat[(idx + 1):(idx + 1 + r * C)])
            idx += 1 + r * C
        X.alpha_set(alpha_pred)
        X.beta_set(beta_pred)
        Q_Q = X.Q_matrix()
        
    alphas.append(alpha_pred)
    betas.append(beta_pred)

    # Make predictions
    forecast_grOU[i, :, :] = np.mean(np.array([simulate_MCAR_compound_poisson([i for i in range(forecast_step + 1)], Q_Q, np.append(window_data[:,-1], [0] * K * (len(R_L) - 1)), b, Sigma, 1, jump_F, uniform=True)[:, 1:] for _ in range(1)]), axis=0)
    df_grOU = np.append(df_grOU, forecast_grOU[i, :, 0:1], axis = 1)

with open(os.path.join('data\\LOBSTERdata\\output\\train\\aggregate', 'forecast_grOU.npy'), 'wb') as f:
            np.save(f, forecast_grOU)

RMSE_grOU = np.sqrt(np.mean((forecast_grOU - test_set) ** 2))
print(f'RMSE for graph OU Edge model is: {RMSE_grOU}')
```

```python
# Univariate AR Edge Model

df_AR = np.array(Y.copy())
aic_values = []
best_p = np.zeros((K, ))
for i in range(K):
    for p in range(1, 6):
        model = AutoReg(df_AR[i, :], lags=p).fit()
        aic_values.append((p, model.aic))
    best_p[i] = min(aic_values, key=lambda x: x[1])[0]
print(f'The best fitted lag for each edge is: {best_p}')

forecast_AR = np.zeros((test_size, K, forecast_step))
for i in range(test_size):
    for j in range(K):
        window_data = df_AR[j, i:]
        model = AutoReg(window_data, lags=best_p[j]).fit()
        forecast_AR[i, j, :] = model.predict(start=len(window_data), end=len(window_data)+forecast_step-1)
    df_AR = np.append(df_AR, forecast_AR[i, :, 0:1], axis = 1)
    
with open(os.path.join('data\\LOBSTERdata\\output\\train\\aggregate', 'forecast_AR.npy'), 'wb') as f:
            np.save(f, forecast_AR)
    
RMSE_AR = np.sqrt(np.mean((forecast_AR - test_set) ** 2))
print(f'RMSE for UAR Edge model is: {RMSE_AR}')
```

```python
# VAR Edge Model

df_VAR = np.array(Y.T.copy())
model = VAR(df_VAR)
# automatic lag selection by AIC
best_lag_var = model.select_order(maxlags=5).aic
print(f'The best fitted lag is: {best_lag_var}')

forecast_VAR = np.zeros((test_size, K, forecast_step))
for i in range(test_size):
    window_data = df_VAR[i:, :]
    model = VAR(window_data).fit(best_lag_var)
    forecast_VAR[i, :, :] = model.forecast(window_data[-model.k_ar:, :], steps=forecast_step).T
    df_VAR = np.append(df_VAR, forecast_VAR[i, :, 0:1].T, axis = 0)

with open(os.path.join('data\\LOBSTERdata\\output\\train\\aggregate', 'forecast_VAR.npy'), 'wb') as f:
            np.save(f, forecast_VAR)

RMSE_VAR = np.sqrt(np.mean((forecast_VAR - test_set) ** 2))
print(f'RMSE for VAR Edge model is: {RMSE_VAR}')
```

```python
# Naive model (pure noise)

forecast_NA = np.zeros((test_size, K, forecast_step))
df_NA = np.array(Y.copy())
NA_noise = np.random.multivariate_normal(mean=np.mean(df_NA, axis=1), cov=np.cov(df_NA, rowvar=True), size=test_size)

for i in range(test_size):
    forecast_NA[i, :, :] = (np.array([NA_noise[i,:]] * forecast_step)).T
    
RMSE_NA = np.sqrt(np.mean((forecast_NA - test_set) ** 2))
print(f'RMSE for Naive model is: {RMSE_NA}')
```

```python
# OU Edge model

forecast_OU = np.zeros((test_size, K, forecast_step))
df_OU = np.array(Y.copy())
Adj_L_OU = [np.zeros((K, K)) for _ in range(max(R_L))]
W_C_OU = edge_weight(Adj_L_OU)
X_OU = GraphOUEdge(R_L, K, C, Adj_L_OU, W_C_OU, 'data\\LOBSTERdata\\output\\train\\aggregate', 'plots')
X_OU.estimate_unknown_Levy(Y, f'{jump_activity}')
b_OU = np.load(f'./data/LOBSTERdata/output/train/aggregate/MCAR_estimates_b_{jump_activity}.npy')
Sigma_OU = np.load(f'./data/LOBSTERdata/output/train/aggregate/MCAR_estimates_Sigma_{jump_activity}.npy')
F_mean_OU = np.load(f'./data/LOBSTERdata/output/train/aggregate/MCAR_estimates_F_mean_{jump_activity}.npy')
F_squared_OU = np.load(f'./data/LOBSTERdata/output/train/aggregate/MCAR_estimates_F_squared_{jump_activity}.npy')

jump_F_OU = scipy.stats.multivariate_normal(mean=F_mean_OU, cov=(F_squared_OU - np.outer(F_mean_OU, F_mean_OU)), allow_singular=True)
alphas_OU = []

for i in range(test_size):
    window_data = df_OU[:, i:]
    theta_hat_OU, HH, H = estimate_OU_drift(window_data, R_L, K, C, P_, Q_, b_OU, None, Sigma_OU, with_cov=True)
    X_OU.alpha_set(theta_hat_OU)
    X_OU.beta_set([[0] * stage * C for stage in R_L])
    try:
        Q_Q = X_OU.Q_matrix()
    except:
        # for l = 1 or 2
        if len(R_L) < 3:
            theta_hat_OU = minimize(lambda x: -np.matmul(x, H) + 0.5 * np.matmul(x, np.matmul(HH, x)), theta_hat_OU,
                                    constraints=({'type': 'ineq', 'fun': lambda x: x}), method='SLSQP').x
        # for l >= 3
        else:
            theta_hat_OU = minimize(lambda x: -np.matmul(x, H) + 0.5 * np.matmul(x, np.matmul(HH, x)), theta_hat_OU,
                                    constraints=({'type': 'ineq', 'fun': lambda x: hurwitz_principal_minors(x)},
                                                 {'type': 'ineq', 'fun': lambda x: x}), method='SLSQP').x
        theta_hat_OU = np.clip(theta_hat_OU, 1e-16, None)
        X_OU.alpha_set(theta_hat_OU)
        X_OU.beta_set([[0] * stage * C for stage in R_L])
        Q_Q = X_OU.Q_matrix()
        
    alphas_OU.append(theta_hat_OU)
    forecast_OU[i, :, :] = np.mean(np.array([simulate_MCAR_compound_poisson([i for i in range(forecast_step + 1)], Q_Q, np.append(window_data[:,-1], [0] * K * (len(R_L) - 1)), b_OU, Sigma_OU, 1, jump_F_OU, uniform=True)[:, 1:] for _ in range(1)]), axis=0)   
    df_OU = np.append(df_OU, forecast_OU[i, :, 0:1], axis = 1)

with open(os.path.join('data\\LOBSTERdata\\output\\train\\aggregate', 'forecast_OU.npy'), 'wb') as f:
            np.save(f, forecast_OU)
    
RMSE_OU = np.sqrt(np.mean((forecast_OU - test_set) ** 2))
print(f'RMSE for OU Edge model is: {RMSE_OU}')
```
