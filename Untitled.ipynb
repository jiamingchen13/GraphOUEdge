```python
from data_process import *
from mcar.estimate import *
from mcar.simulate import *
from graphOUedge import GraphOUEdge

# Initialization
# Tickers
TICKERS = ['SPY', 'AMZN', 'JPM', 'GS', 'KO', 'MSFT', 'NDAQ', 'MNST']

# Number of covariate
C = 1

# Aggregation interval for realized correlation/covariance (1 hour)
rule = 'h'

# Sample interval (1 second)
sample_interval = '1s'

# Forecast steps
forecast_step = 1

# Set hyperparameter gamma
gamma = 1.01
```

```python
# Data Processing
# Training set
for TICKER in TICKERS:
    input_path = './data/LOBSTERdata/_' + TICKER + '_2023-01-01_2025-06-30'
    log_path = './data/LOBSTERdata/log/train/' + TICKER + '_processing_logs'
    os.makedirs(log_path, exist_ok=True)
    output_path = './data/LOBSTERdata/output/train/' + TICKER
    os.makedirs(output_path, exist_ok=True)
    multiprocess_orderbooks(TICKER=TICKER, input_path=input_path, output_path=output_path, log_path=log_path, check_for_processed_data=True, sample_interval = sample_interval)

aggregate_assets(TICKERS, './data/LOBSTERdata/output/train/', './data/LOBSTERdata/output/train/aggregate/')

# Testing set
for TICKER in TICKERS:
    input_path = './data/LOBSTERdata/_' + TICKER + '_2025-07-01_2026-02-01'
    log_path = './data/LOBSTERdata/log/test/' + TICKER + '_processing_logs'
    os.makedirs(log_path, exist_ok=True)
    output_path = './data/LOBSTERdata/output/test/' + TICKER
    os.makedirs(output_path, exist_ok=True)
    multiprocess_orderbooks(TICKER=TICKER, input_path=input_path, output_path=output_path, log_path=log_path, check_for_processed_data=True, sample_interval = sample_interval)
    
aggregate_assets(TICKERS, './data/LOBSTERdata/output/test/', './data/LOBSTERdata/output/test/aggregate/')
```

```python
# Simulation study
# Setup parameters for simulated data
adj_sim = adj_to_line_graph([[0,1,0],[1,0,1],[0,1,0]], ['a','b','c'])
W_C_sim = edge_weight(adjacency_matrices(adj_sim[0], 1))
X_sim = GraphOUEdge([1,1], 2, 1, adjacency_matrices(adj_sim[0], 1), W_C_sim, gamma, 'data\\LOBSTERdata\\output\\sim')
X_sim_sol = GraphOUEdge([1,1], 2, 1, adjacency_matrices(adj_sim[0], 1), W_C_sim, gamma, 'data\\LOBSTERdata\\output\\sim')
X_sim.alpha_set(np.array([[4,2],[2,1]]))
X_sim.beta_set([[1],[1]])
X_sim.Q_matrix()
estimate = []
```

```python
for j in range(25):
    X_sim.simulate('CP',40, 1)

    with open('./data/LOBSTERdata/output/sim/MCAR_CP.npy', 'rb') as f:
        sim = np.load(f)
    with open('./data/LOBSTERdata/output/sim/MCAR_CP_' + str(j) + '.npy', 'wb') as f:
        np.save(f, sim)
    estimates_sample = np.zeros((40, 3, 6))
    P = np.linspace(0, 8, 8 ** 7 + 1)
    for i in range(40):
        for n in range(1, 4):
            T = 2 ** n
            subset_P = slice(0, int(8 ** 7 / 2 ** (3 - n)) + 1, 2 ** (6 * (3 - n)))
            P_ = P[subset_P]  # delta_P = T^(-6)
            Q_ = P_[::T ** 4]  # delta_Q = T^(-2)
            Y_ = sim[i][:, subset_P]

            b_sim_sol, Sigma_sim_sol = X_sim_sol.estimate_unknown_Levy(Y_, P_, Q_, 'grOU')
            theta_hat_sim_sol, _, _ = estimate_grOU_drift(Y_, adjacency_matrices(adj_sim[0], 1), W_C_sim, [1, 1], 1, P_, Q_, b_sim_sol, gamma, None, Sigma_sim_sol, with_cov=True, is_global=False)
            estimates_sample[i, n - 1, :] = theta_hat_sim_sol.reshape(-1)
    estimate.append(estimates_sample)

with open('./data/LOBSTERdata/output/sim/estimate.npy', 'wb') as f:
    np.save(f, np.concatenate(estimate, axis=0))
```

```python
with open('./data/LOBSTERdata/output/sim/estimate.npy', 'rb') as f:
    estimate = np.load(f)

fig, axes = plt.subplots(3, 3, figsize=(24, 18), squeeze=False)
for log2t in range(1, 4):
    for n in range(2):
        for m in range(1):
            ax = axes[n, log2t - 1]
            x = estimate[:, log2t - 1, n * 3 + m]
            x = x[abs(x) < 1e3]
            y = estimate[:, log2t - 1, n * 3 + m + 1]
            y = y[abs(y) < 1e3]

            k = scipy.stats.gaussian_kde(np.vstack([x, y]))
            xi, yi = np.mgrid[0:8:100 * 1j, 0:8:100 * 1j]
            zi = k(np.vstack([xi.flatten(), yi.flatten()]))

            ax.contourf(xi, yi, zi.reshape(xi.shape), cmap='Blues_r')

            ax.set_xticks([0, 1, 2, 3, 4, 5, 6, 7, 8])
            ax.set_yticks([0, 1, 2, 3, 4, 5, 6, 7, 8])
            ax.set_xlabel(r'$\alpha_{%i,%i}$' % (2 * m + 1, n + 1), fontsize=15)
            ax.set_ylabel(r'$\alpha_{%i,%i}$' % (2 * m + 2, n + 1), fontsize=15)
            ax.plot(X_sim.Q_get()[n][2 * m][2 * m], X_sim.Q_get()[n][2 * m + 1][2 * m + 1], color='red', marker='.', markersize=15)
            ax.set_box_aspect(1)

            if n == 0 and m == 0:
                ax.set_title(r'$t = %i$' % 2 ** log2t, fontsize=20)

    ax = axes[2, log2t - 1]
    x = estimate[:, log2t - 1, 2]
    x = x[abs(x) < 1e3]
    y = estimate[:, log2t - 1, 5]
    y = y[abs(y) < 1e3]

    k = scipy.stats.gaussian_kde(np.vstack([x, y]))
    xi, yi = np.mgrid[-3:6:100 * 1j, -3:6:100 * 1j]
    zi = k(np.vstack([xi.flatten(), yi.flatten()]))

    ax.contourf(xi, yi, zi.reshape(xi.shape), cmap='Blues_r')

    ax.set_xticks([-3,-2,-1, 0, 1, 2, 3, 4, 5, 6])
    ax.set_yticks([-3,-2,-1, 0, 1, 2, 3, 4, 5, 6])
    ax.set_xlabel(r'$\beta_{%i}$' % 1, fontsize=15)
    ax.set_ylabel(r'$\beta_{%i}$' % 2, fontsize=15)
    ax.plot(1, 1, color='red', marker='.', markersize=15)
    ax.set_box_aspect(1)

fig.savefig('./data/LOBSTERdata/output/sim/estimate.eps', bbox_inches='tight', format='eps')
```

```python
with open('./data/LOBSTERdata/output/sim/MCAR_CP.npy', 'rb') as f:
    sim = np.load(f)

P_ = np.linspace(0, 8, 8 ** 7 + 1)
subset_P = slice(0, int(8 ** 7 / 2) + 1, 2 ** 6)
P_ = P_[subset_P]
alphas_grOU = []
sample_index = 0

# Underfitting
for j in range(9):
    Q_ = np.append(P_[::2 ** j], P_[-1]) if (sim.shape[2] - 1) % 2 ** j != 0 else P_[::2 ** j]
    X_grOU = GraphOUEdge([1], 2, 1, adjacency_matrices(adj_sim[0], 1), W_C_sim, gamma, 'data\\LOBSTERdata\\output\\train\\aggregate')
    b_grOU, Sigma_grOU = X_grOU.estimate_unknown_Levy(sim[sample_index][:,subset_P], P_, Q_, 'grOU')
    theta_hat_grOU, _, _ = estimate_grOU_drift(sim[sample_index][:,subset_P], adjacency_matrices(adj_sim[0], 1), W_C_sim, [1], 1, P_, Q_, b_grOU, gamma, None, Sigma_grOU, with_cov=True, is_global=False)
    alphas_grOU.append(theta_hat_grOU.reshape(-1))

df_1 = pd.DataFrame(np.array(alphas_grOU))
with open('./data/LOBSTERdata/output/sim/sim_param_1.npy', 'wb') as f:
    np.save(f, df_1)

alphas_grOU = []
# Match fitting
for j in range(9):
    Q_ = np.append(P_[::2 ** j], P_[-1]) if (sim.shape[2] - 1) % 2 ** j != 0 else P_[::2 ** j]
    X_grOU = GraphOUEdge([1,1], 2, 1, adjacency_matrices(adj_sim[0], 1), W_C_sim, gamma, 'data\\LOBSTERdata\\output\\train\\aggregate')
    b_grOU, Sigma_grOU = X_grOU.estimate_unknown_Levy(sim[sample_index][:,subset_P], P_, Q_, 'grOU')
    theta_hat_grOU, _, _ = estimate_grOU_drift(sim[sample_index][:,subset_P], adjacency_matrices(adj_sim[0], 1), W_C_sim, [1,1], 1, P_, Q_, b_grOU, gamma, None, Sigma_grOU, with_cov=True, is_global=False)
    alphas_grOU.append(theta_hat_grOU.reshape(-1))

df_2 = pd.DataFrame(np.array(alphas_grOU))
with open('./data/LOBSTERdata/output/sim/sim_param_2.npy', 'wb') as f:
    np.save(f, df_2)

alphas_grOU = []
# Overfitting
for j in range(9):
    Q_ = np.append(P_[::2 ** j], P_[-1]) if (sim.shape[2] - 1) % 2 ** j != 0 else P_[::2 ** j]
    X_grOU = GraphOUEdge([1,1,1], 2, 1, adjacency_matrices(adj_sim[0], 1), W_C_sim, gamma, 'data\\LOBSTERdata\\output\\train\\aggregate')
    b_grOU, Sigma_grOU = X_grOU.estimate_unknown_Levy(sim[sample_index][:,subset_P], P_, Q_, 'grOU')
    theta_hat_grOU, _, _ = estimate_grOU_drift(sim[sample_index][:,subset_P], adjacency_matrices(adj_sim[0], 1), W_C_sim, [1,1,1], 1, P_, Q_, b_grOU, gamma, None, Sigma_grOU, with_cov=True, is_global=False)
    alphas_grOU.append(theta_hat_grOU.reshape(-1))

df_3 = pd.DataFrame(np.array(alphas_grOU))
with open('./data/LOBSTERdata/output/sim/sim_param_3.npy', 'wb') as f:
    np.save(f, df_3)

# Plot MLE versus different Delta Q scales
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharex=True)

# First subplot: Underfitting
axes[0].plot(range(9), df_1[0])
axes[0].plot(range(9), [df_1[0][0] * (2 ** i) for i in range(9)], color="red", linestyle="--", linewidth=2, label="Benchmark")
axes[0].legend()
axes[0].set_title('Underfitting')

# Middle subplot: Correct fitting
axes[1].plot(range(9), df_2[0])
axes[1].plot(range(9), [df_2[0][0]] * 9, color="red", linestyle="--", linewidth=2, label="Benchmark")
axes[1].legend()
axes[1].set_title('Correct Fitting')

# Last subplot: Overfitting
axes[2].plot(range(9), df_3[0])
axes[2].plot(range(9), [df_3[0][0] / (2 ** i) for i in range(9)], color="red", linestyle="--", linewidth=2, label="Benchmark")
axes[2].legend()
axes[2].set_title('Overfitting')

# Shared axis labels
fig.supxlabel('Log2 Delta Q')
fig.supylabel('MLE Parameter')

plt.tight_layout()
plt.show()

fig.savefig('./data/LOBSTERdata/output/sim/fitting.eps', bbox_inches='tight', format='eps')
```
